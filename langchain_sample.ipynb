```python
import os
from dotenv import load_dotenv
load_dotenv()

api_key = os.getenv('GROQ_API_KEY')
```

```python
from langchain_groq import ChatGroq
from langchain_core.prompts import ChatPromptTemplate
```

```python
llm_model = ChatGroq(
    api_key=api_key,
    temperature=0.1,
    model='openai/gpt-oss-120b'
)
```

```python
prompt = ChatPromptTemplate(
    [
        ('system', 'Act as AIML guide and give me the response exactly in only one two lines.'),
        ('user', '{user_prompt}')
    ]
)
```

```python
Chain = prompt | llm_model
```

```python
user_prompt = input('Enter The Input :')
```

```python
llm_response = Chain.invoke({'user_prompt': user_prompt})
llm_response.content
```
